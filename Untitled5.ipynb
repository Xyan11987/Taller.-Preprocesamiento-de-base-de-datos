Limpiar y corregir los problemas encontrados

```python

df_limpio = df.copy()

# Convertir a números, detectando errores de texto
for columna in df_limpio.columns:
    df_limpio[columna] = pd.to_numeric(
        df_limpio[columna],
        errors="coerce"
    )

print("Faltantes después de la conversión:")
print(df_limpio.isnull().sum())
```

```python

# Categorías aceptadas para este archivo
categorias = {
    "sex": [0, 1],
    "cp": [0, 1, 2, 3],
    "fbs": [0, 1],
    "restecg": [0, 1, 2],
    "exang": [0, 1],
    "slope": [0, 1, 2],
    "ca": [0, 1, 2, 3],
    "thal": [1, 2, 3],
    "target": [0, 1]
}

for columna, permitidos in categorias.items():
    invalidos = (
        df_limpio[columna].notna()
        & ~df_limpio[columna].isin(permitidos)
    )

    print(columna, "inválidos:", invalidos.sum())
    df_limpio.loc[invalidos, columna] = np.nan

# Edades negativas o superiores a 100 años:
# se marcan para imputación por falta de validación
df_limpio.loc[
    (df_limpio["age"] < 0) |
    (df_limpio["age"] > 100),
    "age"
] = np.nan
```

elimina los duplicados exactos y los registros sin variable objetivo

```python

# Eliminar registros exactamente duplicados
df_limpio = df_limpio.drop_duplicates()

# Eliminar registros sin target
df_limpio = df_limpio.dropna(subset=["target"])

print("Dimensiones después de la limpieza:")
print(df_limpio.shape)

print("\nValores faltantes restantes:")
print(df_limpio.isnull().sum())
```

Detectar y analizar outliers

```python


# Variables numéricas
numericas = [
    "age", "trestbps", "chol",
    "thalach", "oldpeak"
]

# Crear gráficos de caja
plt.figure(figsize=(12, 8))

for i, columna in enumerate(numericas, 1):
    plt.subplot(2, 3, i)
    plt.boxplot(df_limpio[columna].dropna())
    plt.title(columna)

plt.tight_layout()
plt.show()

```

Detectar outliers mediante IQR

```python

for columna in numericas:
    q1 = df_limpio[columna].quantile(0.25)
    q3 = df_limpio[columna].quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    outliers = df_limpio[
        (df_limpio[columna] < limite_inferior) |
        (df_limpio[columna] > limite_superior)
    ]

    print(f"{columna}: {len(outliers)} outliers")
    print("Límites:", limite_inferior, limite_superior)
```

Separar variables predictoras y variable objetivo

```python

# Variables predictoras
X = df_limpio.drop(columns=["target"])

# Variable objetivo
y = df_limpio["target"].astype(int)

print("Variables predictoras:")
print(X.columns)

print("\nVariable objetivo:")
print(y.head())
```

Dividir los datos en entrenamiento y prueba

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

Comprobar la distribución de las clases

```python

print("Distribución original:")
print(y.value_counts(normalize=True))

print("\nDistribución de entrenamiento:")
print(y_train.value_counts(normalize=True))

print("\nDistribución de prueba:")
print(y_test.value_counts(normalize=True))
```

Los datos fueron divididos en un conjunto de entrenamiento del 80 % y un conjunto de prueba del 20 %. Se utilizó una división estratificada para mantener aproximadamente la proporción de pacientes de cada clase. Además, se estableció una semilla aleatoria para garantizar la reproducibilidad.

Tratar valores faltantes, codificar y escalar

```python

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

# Grupos de variables
numericas = [
    "age", "trestbps", "chol",
    "thalach", "oldpeak"
]

binarias = ["sex", "fbs", "exang"]

categoricas = ["cp", "restecg", "thal"]

ordinales = ["slope", "ca"]

# Numéricas: mediana y escalamiento
pipeline_numerico = Pipeline([
    ("imputar", SimpleImputer(strategy="median")),
    ("escalar", StandardScaler())
])

# Binarias y ordinales: moda
pipeline_discreto = Pipeline([
    ("imputar", SimpleImputer(strategy="most_frequent"))
])

# Categóricas: moda y One-Hot
pipeline_categorico = Pipeline([
    ("imputar", SimpleImputer(strategy="most_frequent")),
    ("codificar", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocesador = ColumnTransformer([
    ("numericas", pipeline_numerico, numericas),
    ("binarias", pipeline_discreto, binarias),
    ("categoricas", pipeline_categorico, categoricas),
    ("ordinales", pipeline_discreto, ordinales)
])
```

Aplicar las transformaciones correctamente

```python

# Ajustar solo con entrenamiento
X_train_procesado = preprocesador.fit_transform(X_train)

# Transformar prueba sin volver a ajustar
X_test_procesado = preprocesador.transform(X_test)

print("Train:", X_train_procesado.shape)
print("Test:", X_test_procesado.shape)
```

Comprobar resultados y guardar archivos

```python

# Obtener los nombres de las columnas
columnas_finales = preprocesador.get_feature_names_out()

# Construir tablas
train_final = pd.DataFrame(
    X_train_procesado,
    columns=columnas_finales,
    index=X_train.index
)

test_final = pd.DataFrame(
    X_test_procesado,
    columns=columnas_finales,
    index=X_test.index
)

# Agregar target
train_final["target"] = y_train
test_final["target"] = y_test

# Comprobar valores faltantes
print("Faltantes Train:",
      train_final.isna().sum().sum())

print("Faltantes Test:",
      test_final.isna().sum().sum())

display(train_final.head())

# Guardar CSV procesados
train_final.to_csv("heart_train_limpio.csv", index=False)
test_final.to_csv("heart_test_limpio.csv", index=False)
```
